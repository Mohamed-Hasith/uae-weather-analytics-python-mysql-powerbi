```python
import requests

# Our four UAE cities
cities = {
    "Ajman": {"latitude": 25.4052, "longitude": 55.5136},
    "Dubai": {"latitude": 25.2048, "longitude": 55.2708},
    "Abu Dhabi": {"latitude": 24.4539, "longitude": 54.3773},
    "Sharjah": {"latitude": 25.3463, "longitude": 55.4209}
}

# Open-Meteo API endpoint
url = "https://api.open-meteo.com/v1/forecast"

# Weather variables
params = {
    "latitude": ",".join(str(city["latitude"]) for city in cities.values()),
    "longitude": ",".join(str(city["longitude"]) for city in cities.values()),
    "current": "temperature_2m,relative_humidity_2m,apparent_temperature,precipitation,wind_speed_10m,wind_direction_10m",
    "timezone": "Asia/Dubai"
}

response = requests.get(url, params=params)

print("Status Code:", response.status_code)

weather_data = response.json()

print(weather_data)
```

```python
response = requests.get(url, params=params, timeout=30)

# Check HTTP status
if response.status_code == 200:

    print("API request successful!")

    weather_data = response.json()

    print("Number of locations:", len(weather_data))

else:

    print("API request failed!")
    print("Status Code:", response.status_code)
    print("Error:", response.text)
```

```python
weather_records = []

for city_name, city_data in zip(cities.keys(), weather_data):

    record = {
        "city": city_name,
        "latitude": city_data["latitude"],
        "longitude": city_data["longitude"],
        "time": city_data["current"]["time"],
        "temperature": city_data["current"]["temperature_2m"],
        "humidity": city_data["current"]["relative_humidity_2m"],
        "feels_like": city_data["current"]["apparent_temperature"],
        "precipitation": city_data["current"]["precipitation"],
        "wind_speed": city_data["current"]["wind_speed_10m"],
        "wind_direction": city_data["current"]["wind_direction_10m"]
    }

    weather_records.append(record)

print(weather_records)
```

```python
df = pd.DataFrame(weather_records)

display(df)
```

```python
# Check DataFrame structure
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated(subset=["city", "time"]).sum())
```

```python
# Convert time column from string to datetime
df["time"] = pd.to_datetime(df["time"])

# Check the updated data types
print(df.dtypes)
```

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS weather_observations (

    id BIGINT AUTO_INCREMENT PRIMARY KEY,

    city VARCHAR(50) NOT NULL,

    latitude DECIMAL(10,6),

    longitude DECIMAL(10,6),

    time DATETIME NOT NULL,

    temperature DECIMAL(5,2),

    humidity INT,

    feels_like DECIMAL(5,2),

    precipitation DECIMAL(6,2),

    wind_speed DECIMAL(6,2),

    wind_direction INT,

    CONSTRAINT unique_city_time UNIQUE (city, time)

);
"""

with engine.begin() as connection:
    connection.execute(text(create_table_query))

print("Weather observations table created successfully!")
```

```python
df.to_sql(
    name="weather_observations",
    con=engine,
    if_exists="append",
    index=False
)

print("Weather data inserted successfully!")
```

```python
upsert_query = """
INSERT INTO weather_observations (
    city,
    latitude,
    longitude,
    time,
    temperature,
    humidity,
    feels_like,
    precipitation,
    wind_speed,
    wind_direction
)
VALUES (
    :city,
    :latitude,
    :longitude,
    :time,
    :temperature,
    :humidity,
    :feels_like,
    :precipitation,
    :wind_speed,
    :wind_direction
) AS new

ON DUPLICATE KEY UPDATE

    latitude = new.latitude,
    longitude = new.longitude,
    temperature = new.temperature,
    humidity = new.humidity,
    feels_like = new.feels_like,
    precipitation = new.precipitation,
    wind_speed = new.wind_speed,
    wind_direction = new.wind_direction;
"""

with engine.begin() as connection:

    connection.execute(
        text(upsert_query),
        df.to_dict(orient="records")
    )

print("Weather data inserted/updated successfully!")
```
